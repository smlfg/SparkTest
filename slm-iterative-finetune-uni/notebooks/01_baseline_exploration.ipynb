## Baseline Evaluation laden (wenn vorhanden)

```python
import os

eval_file = '../data/eval_logs/iteration_0_eval.json'

if os.path.exists(eval_file):
    results = load_json(eval_file)
    summary = results['summary']
    
    print_section("Baseline Evaluations-Ergebnisse")
    print(f"Pass Rate: {summary['pass_rate']*100:.1f}%")
    print(f"Syntax korrekt: {summary['syntax_valid_rate']*100:.1f}%")
    print(f"Hat Funktion: {summary['has_function_rate']*100:.1f}%")
    print(f"\nFehlertypen:")
    for error_type, count in summary['error_types'].items():
        print(f"  {error_type}: {count}")
else:
    print("⚠️  Noch keine Baseline-Evaluation vorhanden.")
    print("   Führe aus: python ../src/eval.py --iteration 0 --model <base_model>")
```

## Fehlerbeispiele ansehen

```python
if os.path.exists(eval_file):
    failed_results = [r for r in results['results'] if not r.get('passed', False)]
    
    print(f"\n❌ {len(failed_results)} fehlgeschlagene Aufgaben\n")
    
    # Erste 3 Fehler ansehen
    for i, result in enumerate(failed_results[:3], 1):
        print(f"\n{'='*80}")
        print(f"Fehlerbeispiel {i}: {result['task_id']}")
        print(f"{'='*80}")
        print(f"\n📝 Prompt:")
        print(result['prompt'][:200] + "...")
        print(f"\n💻 Generierter Code:")
        print(result['generated_code'])
        print(f"\n❌ Fehler: {result.get('error', 'Unknown')}")
        print(f"   Fehlertyp: {result.get('error_type', 'Unknown')}")
```

## Visualisierung

```python
if os.path.exists(eval_file):
    # Fehlertypen visualisieren
    error_types = summary['error_types']
    
    plt.figure(figsize=(10, 6))
    plt.bar(error_types.keys(), error_types.values(), color='coral')
    plt.title('Fehlertypen in Baseline (Iteration 0)', fontsize=14, fontweight='bold')
    plt.xlabel('Fehlertyp')
    plt.ylabel('Anzahl')
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()
```
